1.Is the average final GPA of the participants greater than
 3.3
?


```python
# Final GPA
final_gpa = df["final_gpa"].dropna()

# One-sample t-test against 3.3
t_stat, p_two = stats.ttest_1samp(final_gpa, popmean=3.3)

alpha = 0.05
df_gpa = len(final_gpa) - 1

# Convert to one-tailed p-value and calculate critical t-value for right-tailed test
if t_stat > 0:
    p_one = p_two / 2
    critical_t_value = t.ppf(1 - alpha, df_gpa)
else:
    p_one = 1 - (p_two / 2)
    critical_t_value = t.ppf(alpha, df_gpa) # Though this case is less likely for 'greater than' hypothesis

# Sample statistics
results = pd.DataFrame({
    "Statistic":[
        "Sample Size",
        "Sample Mean",
        "Sample Standard Deviation",
        "Hypothesized Mean",
        "Critical t-value",
        "t Statistic",
        "One-tailed p-value"
    ],
    "Value":[
        len(final_gpa),
        final_gpa.mean(),
        final_gpa.std(),
        3.3,
        critical_t_value,
        t_stat,
        p_one
    ]
})

display(results.round(4))
```

Here is the evaluation for the first question.

**Q1. Is the average final GPA of the participants greater than 3.3?**

1. Null Hypothesis ($H_0$)
$H_0:\mu\le3.3$
(The average final GPA of the population is less than or equal to 3.3.)
2. Alternative Hypothesis ($H_a$)
$H_a:\mu>3.3$
(The average final GPA of the population is greater than 3.3.)
3. Significance Level ($\alpha$)
$\alpha=0.05$
4. Test Statistic
Using the formula for the one-sample t-statistic with sample standard deviation ($s$):
$t=\frac{\bar{x}-\mu_0}{s/\sqrt{n}}$
Substituting the provided summary values:

* Sample Size ($n$) = 95
* Sample Mean ($\bar{x}$) = 3.3254
* Sample Standard Deviation ($s$) = 0.4374
* Hypothesized Mean ($\mu_0$) = 3.3000

$t=\frac{3.3254-3.3000}{0.4374/\sqrt{95}} \approx 0.5653$
Test Statistic ($t$): 0.5653

5. Critical Value
For a right-tailed t-test at $\alpha=0.05$:
Critical Value ($t_{critical}$): 1.6612
6. P-Value
One-tailed P-value: 0.2866
7. Decision
Fail to reject $H_0$ (Do not reject the null hypothesis).

* P-value rule: $P\text{-value}=0.2866 > \alpha=0.05$.
* Critical value rule: $t=0.5653 < t_{critical}=1.6612$ (the test statistic does not fall into the critical rejection region).

8. Conclusion
Connection of P-Value against $H_0$: Because the P-value (0.2866) is greater than the significance level ($\alpha=0.05$), we fail to reject $H_0$. This means that the probability of obtaining a sample mean of 3.3254 or higher purely by random chance when the true population mean is 3.3 is roughly 28.66%. Since this chance is well above our 5% threshold, there is no statistically significant evidence to support $H_a$.
Verdict: No, we cannot conclude that the average final GPA of the participants is greater than 3.3.

### Critical Z-value Calculation for One-tailed Tests

```python
import numpy as np
from scipy.stats import norm

alpha = 0.05

# For a right-tailed test (e.g., to see if a mean is GREATER than a hypothesized value)
# The critical value is at 1 - alpha
right_tailed_critical_z = norm.ppf(1 - alpha)

# For a left-tailed test (e.g., to see if a mean is LESS than a hypothesized value)
# The critical value is at alpha
left_tailed_critical_z = norm.ppf(alpha)

print(f"For alpha = {alpha} (one-tailed test):")
print(f"Right-tailed critical Z-value (for H1: μ > μ0): {right_tailed_critical_z:.4f}")
print(f"Left-tailed critical Z-value (for H1: μ < μ0): {left_tailed_critical_z:.4f}")
```
