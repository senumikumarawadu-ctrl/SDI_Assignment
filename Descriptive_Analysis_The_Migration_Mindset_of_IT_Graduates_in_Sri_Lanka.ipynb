```python
#Numerical variable 1 - Age
print("Summary Statistics for Age:")
col = df["age"]

n = col.count()
mean = col.mean()
median = col.median()
mode = col.mode()[0]
minimum = col.min()
maximum = col.max()
range_ = maximum - minimum
variance = col.var()
std_dev = col.std()
skewness = col.skew()
kurtosis = col.kurt()

print(f"n={n}, mean={mean:.2f}, median={median:.2f}, mode={mode}")
print(f"min={minimum}, max={maximum}, range={range_}")
print(f"variance={variance:.2f}, std_dev={std_dev:.2f}")
print(f"skewness={skewness:.2f}, kurtosis={kurtosis:.2f}")
```

```python
age = df["age"]

print("\nSummary Statistics for Age:")
display(age.describe())
```
