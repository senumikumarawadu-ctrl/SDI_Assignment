Can final GPA be predicted using age, number of dependents,experience,number of certifications,number of relatives living abroad and number of close friends in IT job role who have already migrated?

```python
import statsmodels.api as sm
df_cleaned = df.dropna(subset=['final_gpa'])
# Independent variables
X = df_cleaned[[
    "age",
    "num_of_dep",
    "experience",
    "num_certs",
    "relatives_abroad",
    "friends_it_migrated"
]]

# Dependent variable
y = df_cleaned["final_gpa"]

# Add intercept
X = sm.add_constant(X)

# Fit regression model
model = sm.OLS(y, X).fit()

# Display regression summary
print(model.summary())
```

```python
# Predicted values
y_pred = model.predict(X)

# Residuals
residuals = y - y_pred
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=y,
    y=y_pred,
    color=sns.color_palette("viridis")[4],
    s=70,
    alpha=0.8
)

# Perfect prediction line
plt.plot(
    [y.min(), y.max()],
    [y.min(), y.max()],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.title("Actual vs. Predicted Final GPA", fontsize=18, weight="bold")
plt.xlabel("Actual Final GPA", fontsize=13)
plt.ylabel("Predicted Final GPA", fontsize=13)
plt.legend()

plt.tight_layout()
plt.show()


```

## Actual vs. Predicted Final GPA

- **Weak-to-moderate fit:** The fit is weak to moderate. The points trend upward. The cloud is wide and flatter than the red 45° line so the predictors explain only part of the variation in GPA.

- **Compressed predictions:** Actual GPA values range from 1.9 to 4.0 but predicted GPA values sit mostly between 3.0 and 3.6. The model pulls everyone toward the average.

- **Bias at the extremes:** Low-GPA respondents (about 1.9 to 2.1) are over‑predicted by more than a point and the 4.0 respondent is under‑predicted (about 3.45).

- **Outliers:** The few respondents near 2.0 are far from the rest. May be pulling the fitted line. Check their influence (Cooks distance). Confirm the values are correct.

- **Conclusion:** The model shows an upward trend but is not reliable, for predicting an individuals GPA.




```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=y_pred,
    y=residuals,
    color=sns.color_palette("viridis")[4],
    s=70,
    alpha=0.8
)

plt.axhline(
    y=0,
    color="red",
    linestyle="--",
    linewidth=2
)

plt.title("Residuals vs. Predicted Values",
          fontsize=18,
          weight="bold")
plt.xlabel("Predicted Final GPA", fontsize=13)
plt.ylabel("Residuals", fontsize=13)

plt.tight_layout()
plt.show()
```

## Residuals vs. Predicted Final GPA

- **No curve or funnel:** Most residuals stay close to zero. Spread out evenly which suggests the assumptions of linearity and constant variance are probably okay.

- **Three large outliers:** There are three residuals that stand out: around -1.1, -1.3 and -1.4 all at predicted values between 3.2 and 3.4. These correspond to GPAs around 1.9 to 2.1 so the model predicted much higher than what was observed. These points are far from the rest.

- **Skew toward residuals:** All the large errors are negative. The largest positive residuals are about +0.6. This imbalance means the residual distribution likely has a tail on the left side, which suggests the normality assumption is probably not met.

- **Typical error:** Most residuals are between -0.5 and +0.6 GPA points. So for a student the model’s prediction is off by about half a point or so.

- **Limited view at the ends:** The predicted values are mostly between 3.0 and 3.6 with few below 3.0 or above 3.7. Because of this we can’t rule out the possibility of variance, at the extremes.

- **Conclusion.** The model works well for most students: Those three extreme outliers are pulling the fit down. It’s worth checking how influential those points are using Cook’s distance. Consider refitting the model without those points or trying regression to see if the model results change.


```python
plt.figure(figsize=(8,6))

sns.histplot(
    residuals,
    kde=True,
    color=sns.color_palette("viridis")[4]
)

plt.title("Distribution of Residuals",
          fontsize=18,
          weight="bold")
plt.xlabel("Residuals", fontsize=13)
plt.ylabel("Frequency", fontsize=13)

plt.tight_layout()
plt.show()
```

## Distribution of Residuals

- **Left-skewed, not bell-shaped:** I see the tallest bar is to the right of zero around +0.1 to +0.27 with about 28 respondents. A long tail stretches to the left.

- **Outlier tail:** I see three residuals, between -1.35 and -1.0. Those are the same low‑GPA respondents that appeared in plots and they create the skew.

- **Bulk of the data:** I see most residuals, 75 out of 95 lie between -0.45 and +0.46. This range is fairly symmetric. The skew comes mainly from an extreme cases.

- **Normality is likely violated:** I see a Shapiro‑Wilk test would probably reject normality and a Q‑Q plot would show points dropping from the line at the lower end.

- **Impact:** I see coefficient estimates stay unbiased. With 95 observations the central limit theorem helps a bit.. P‑values and confidence intervals may be unreliable because outliers raise the spread of errors.

- **Fix:** I see the first step is to confirm low‑GPA values are correct. Then refit without those cases. Use robust regression. Check if the conclusions change.

**Residuals vs. Predicted**

- **U-shaped pattern:** The residuals are positive when predictions are low (below 2) mostly negative when predictions are in the middle (2 to 12) and positive again when predictions are high (above 15). This shows the relationship between age and experience is not straight.

- **Mild funnel:** The spread increases when predictions are higher.

- **Outlier:** One person has a residual of about +8.7, which means their experience is much higher than the model predicts.

**Distribution of Residuals**

- ** bell-shaped with a right skew:** The highest point is below zero and a long tail goes up to about +8.7 so the data does not look normal.

- **Cause:** The skew is, from the few people who have very high experience.

**Conclusion:** Age is a predictor but the linearity assumption is not correct. Adding an age squared term or log-transforming experience could fix the curve the skew and the negative predictions. Also check the person with the residual.



## Regression Model 1: Predicting Final GPA with scikit-learn



```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Drop rows where 'final_gpa' is NaN for consistency with previous analysis
df_cleaned_gpa = df.dropna(subset=['final_gpa'])

# Define features (X) and target (y) for GPA prediction
X_gpa = df_cleaned_gpa[[
    "age",
    "num_of_dep",
    "experience",
    "num_certs",
    "relatives_abroad",
    "friends_it_migrated"
]]
y_gpa = df_cleaned_gpa["final_gpa"]

# Split the data into training and testing sets
X_train_gpa, X_test_gpa, y_train_gpa, y_test_gpa = train_test_split(
    X_gpa, y_gpa, test_size=0.2, random_state=42
)

# Initialize and train the Linear Regression model
model_gpa = LinearRegression()
model_gpa.fit(X_train_gpa, y_train_gpa)

# Make predictions on the test set
y_pred_gpa = model_gpa.predict(X_test_gpa)

# Evaluate the model
mse_gpa = mean_squared_error(y_test_gpa, y_pred_gpa)
r2_gpa = r2_score(y_test_gpa, y_pred_gpa)

print("--- Model Evaluation for Final GPA Prediction ---")
print(f"Mean Squared Error: {mse_gpa:.2f}")
print(f"R-squared: {r2_gpa:.2f}")
print("\nModel Coefficients:")
for feature, coef in zip(X_gpa.columns, model_gpa.coef_):
    print(f"{feature}: {coef:.3f}")
print(f"Intercept: {model_gpa.intercept_:.3f}")
```

### Visualizing Results for GPA Prediction Model

```python
# Actual vs. Predicted values plot
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=y_test_gpa,
    y=y_pred_gpa,
    color=sns.color_palette("viridis")[4],
    s=70,
    alpha=0.8
)

# Perfect prediction line
plt.plot(
    [y_test_gpa.min(), y_test_gpa.max()],
    [y_test_gpa.min(), y_test_gpa.max()],
    color="red",
    linestyle="--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.title("Actual vs. Predicted Final GPA (Scikit-learn)", fontsize=18, weight="bold")
plt.xlabel("Actual Final GPA", fontsize=13)
plt.ylabel("Predicted Final GPA", fontsize=13)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Residuals vs. Predicted Values plot
residuals_gpa = y_test_gpa - y_pred_gpa

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=y_pred_gpa,
    y=residuals_gpa,
    color=sns.color_palette("viridis")[4],
    s=70,
    alpha=0.8
)
plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
plt.title("Residuals vs. Predicted Values (GPA Scikit-learn)", fontsize=18, weight="bold")
plt.xlabel("Predicted Final GPA", fontsize=13)
plt.ylabel("Residuals", fontsize=13)
plt.tight_layout()
plt.show()
```



**1. Linearity**
- **How to verify:** Residuals vs. predicted plot (random band = OK, curve = violated), residuals vs. each predictor, partial regression plots, Ramsey RESET test.
- **GPA model:** Looks acceptable. There is no clear curve.
- **Experience model:** **Violated.** The residuals form a U-shape.
- **Fix:** Add an age² term, or transform the outcome (log).

**2. Independence of observations**
- **How to verify:** Durbin-Watson statistic (about 2 = no autocorrelation), residuals in order, ACF plot. For survey data, the study design matters most.
- **Both models:** Probably satisfied. Each row is a separate respondent, and row order has no meaning, so the tests add little. Independence would fail only if respondents were clustered (same company or referral chain).

**3. Homoscedasticity (constant variance)**
- **How to verify:** Residuals vs. predicted plot (constant-width band = OK, funnel = violated), scale-location plot, Breusch-Pagan test, White test.
- **GPA model:** Reasonably constant, though a few large negative outliers stretch the spread.
- **Experience model:** Mildly violated. Spread grows at higher predicted values.
- **Fix:** Log-transform the outcome, use weighted least squares, or use robust standard errors (HC3).

**4. Normality of residuals**
- **How to verify:** Residual histogram, Q-Q plot (points on the diagonal = OK), Shapiro-Wilk test, Jarque-Bera test.
- **GPA model:** **Likely violated.** There is a long left tail from three extreme low-GPA respondents.
- **Experience model:** Questionable. It has a right skew from a few high-experience respondents.
- **Fix:** Transform the outcome, check or remove outliers, or use bootstrap confidence intervals. With about 95 observations, moderate departures are tolerable.

**5. No multicollinearity (GPA model only)**
- **How to verify:** Correlation matrix and VIF (above about 5 to 10 is a problem).
- **Risk:** `age` and `experience` are probably strongly correlated, so check VIF. This does not apply to the experience model, which has one predictor.
