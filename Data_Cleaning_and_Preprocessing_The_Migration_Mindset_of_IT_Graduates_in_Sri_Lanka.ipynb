```python

import pandas as pd
import numpy as np
import re
file_path = "/content/The_Migration_Mindset_of_IT_Graduates_in_Sri_Lanka.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully")
print("Original shape:", df.shape)

df.head()
```

```python
df = df.drop(columns=["Timestamp"], errors="ignore")

print("Timestamp removed")

df.head()
```

```python
print("Missing values before cleaning:")

print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates removed")
```

```python
df.columns = df.columns.str.strip().str.lower()
print(df.columns.tolist())

print("Column names cleaned")

print(df.dtypes)
numeric_columns = [
    "age",
    "num_of_dep",
    "final_gpa",
    "experience",
    "num_certs",
    "current_job_role_sat_rate",
    "relatives_abroad",
    "friends_it_migrated"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print(df.dtypes)
```

```python
df.columns = df.columns.str.strip()

categorical_columns = [
    "gender",
    "marital_status",
    "breadwinner",
    "uni_type",
    "family_living_abroad",
    "plan_to_migrate"
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts(dropna=False))
```

```python
df[
    [
        "age",
        "final_gpa",
        "experience",
        "num_certs",
        "current_job_role_sat_rate"
    ]
].describe()

df.loc[
    (df["age"] < 18) | (df["age"] > 70),
    "age"
] = np.nan

df.loc[
    (df["final_gpa"] <= 0) | (df["final_gpa"] > 4),
    "final_gpa"
] = np.nan

df.loc[
    (df["experience"] < 0) | (df["experience"] > 50),
    "experience"
] = np.nan

df.loc[
    (df["current_job_role_sat_rate"] < 1) |
    (df["current_job_role_sat_rate"] > 5),
    "current_job_role_sat_rate"
] = np.nan

df.head()
```

```python
df.head(10)
df.info()
print("Missing values after cleaning:")

print(
    df.isnull()
    .sum()
    .sort_values(ascending=False)
)
print("Remaining duplicate rows:")

print(df.duplicated().sum())

print("Final dataset shape:")

print(df.shape)
```

```python
def clean_specialization(value):

    if pd.isna(value):
        return pd.NA

    value_original = str(value).strip().rstrip(".")

    value_clean = re.sub(
        r"\s+",
        " ",
        value_original
    ).casefold()

    specialization_mapping = {

        # Computer Science
        "computer science": "Computer Science",
        "cs": "Computer Science",
        "bsc in computer science": "Computer Science",
        "bachelors of computer science special":
            "Computer Science",

        # Computer Science & Engineering
        "computer science engineering":
            "Computer Science & Engineering",

        "computer science & engineering":
            "Computer Science & Engineering",

        "computer science and engineering":
            "Computer Science & Engineering",

        "computer science engineer":
            "Computer Science & Engineering",

        "cse":
            "Computer Science & Engineering",

        # Software Engineering
        "software engineering":
            "Software Engineering",

        "software engineer":
            "Software Engineering",

        "se bsc":
            "Software Engineering",

        "specialization in software engineering":
            "Software Engineering",

        # Information Technology
        "it":
            "Information Technology",

        "information technology":
            "Information Technology",

        "bsc(it)":
            "Information Technology",

        "it- bsc":
            "Information Technology",

        # Information Systems
        "information systems":
            "Information Systems",

        "computing and information systems":
            "Information Systems",

        # Electronics / Telecommunications
        "electronic and telecommunication engineering":
            "Electronics & Telecommunication Engineering",

        "electronics & telecommunication engineering":
            "Electronics & Telecommunication Engineering",

        # Network
        "network engineer":
            "Network Engineering",

        "network engineering":
            "Network Engineering",

        # Artificial Intelligence
        "ai":
            "Artificial Intelligence",

        "ai engineer":
            "Artificial Intelligence",

        # Correct obvious typo
        "stat and coputer sceience":
            "Statistics and Computer Science",

        # Other case corrections
        "mechanical engineering":
            "Mechanical Engineering",

        "material science and engineering":
            "Material Science and Engineering",

        "computational physics":
            "Computational Physics",

        "management":
            "Management",

        "marketing":
            "Marketing",

        "data analysis":
            "Data Analysis",

        "data engineering":
            "Data Engineering",

        "data science":
            "Data Science",

        "data science and analytics":
            "Data Science and Analytics",

        "cybersecurity":
            "Cybersecurity",

        "network & cyber security":
            "Network & Cyber Security"
    }

    return specialization_mapping.get(
        value_clean,
        value_original
    )


df["academic_spl"] = (
    df["academic_spl"]
    .apply(clean_specialization)
    .astype("string")
)
```

```python
output_file = "/content/final_cleaned_dataset.csv"

df.to_csv(
    output_file,
    index=False
)

print("Final cleaned dataset created successfully.")
print(output_file)

from google.colab import files

files.download(
    "/content/final_cleaned_dataset.csv"
)
```
